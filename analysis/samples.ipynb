```python
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
load_path = Path("//home/fred/Data/foundata/out/nts/2023")
schedules_path = load_path / "activities.csv"
attributes_path = load_path / "attributes_binned.csv"
load_path.exists(), schedules_path.exists(), attributes_path.exists()
```

```python
attributes = pd.read_csv(attributes_path, low_memory=False)
schedules = pd.read_csv(schedules_path, low_memory=False)
```

```python
pid_filter = attributes[
    (attributes.access_egress_distance == "≤3.376")
    & (attributes.hh_zone == "suburban")
].pid
len(pid_filter)
```

```python
print(len(schedules.pid.unique()), len(attributes.pid.unique()))

test_schedules = schedules[schedules.pid.isin(pid_filter)]
test_attributes = attributes[attributes.pid.isin(pid_filter)]
print(len(test_schedules.pid.unique()), len(test_attributes.pid.unique()))
test_schedules.to_csv("../tmp/test_schedules.csv", index=False)
test_attributes.to_csv("../tmp/test_attributes.csv", index=False)

training_schedules = schedules[~schedules.pid.isin(pid_filter)]
training_attributes = attributes[~attributes.pid.isin(pid_filter)]
print(
    len(training_schedules.pid.unique()), len(training_attributes.pid.unique())
)
training_schedules.to_csv("../tmp/training_schedules.csv", index=False)
training_attributes.to_csv("../tmp/training_attributes.csv", index=False)
```

```python
load_path = Path("..") / "tmp"
base_file = load_path / "nts_schedules_2019_2023.csv"
base_attributes = load_path / "nts_attributes_2019_2023.csv"
load_path.exists(), base_file.exists(), base_attributes.exists()
```

```python
full_sample = pd.read_csv(base_file)
base_attributes = pd.read_csv(base_attributes)
full_sample.pid.nunique(), base_attributes.pid.nunique()
```

```python
for name, n in [
    (200, 200000),
    (100, 100000),
    (50, 50000),
    (25, 25000),
    (12, 12500),
    (6, 6250),
]:

    ids = full_sample["pid"].unique()
    ids = np.random.choice(ids, size=n, replace=False)
    sample = full_sample[full_sample["pid"].isin(ids)]
    attribute_sample = base_attributes[base_attributes["pid"].isin(ids)]
    sample.to_csv(
        load_path / f"nts_schedules_2019_2023_{name}.csv", index=False
    )
    attribute_sample.to_csv(
        load_path / f"nts_attributes_2019_2023_{name}.csv", index=False
    )
    print(sample.pid.nunique())
```
