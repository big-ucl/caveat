```python
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
load_path = Path("..") / "tmp"
base_file = load_path / "nts_schedules_2023.csv"
load_path.exists(), base_file.exists()
```

```python
full_sample = pd.read_csv(base_file)
full_sample.pid.nunique()
```

```python
for name, n in [(50, 30000), (25, 15000), (12, 7500), (6, 3750)]:

    ids = full_sample["pid"].unique()
    ids = np.random.choice(ids, size=n, replace=False)
    sample = full_sample[full_sample["pid"].isin(ids)]
    sample.to_csv(load_path / f"nts_schedules_2023_{name}.csv", index=False)
    print(sample.pid.nunique())
```
